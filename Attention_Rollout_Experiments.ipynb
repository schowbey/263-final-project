```python
# ACKNOWLEDGEMENTS:
# Implementation of attention rollout as provided by authors of "Quantifying Attention Flow in Transformers": https://github.com/samiraabnar/attention_flow/
import numpy as np

def compute_joint_attention(att_mat, add_residual=True):
    if add_residual:
        residual_att = np.eye(att_mat.shape[1])[None,...]
        aug_att_mat = att_mat + residual_att
        aug_att_mat = aug_att_mat / aug_att_mat.sum(axis=-1)[...,None]
    else:
       aug_att_mat =  att_mat

    joint_attentions = np.zeros(aug_att_mat.shape)

    layers = joint_attentions.shape[0]
    joint_attentions[0] = aug_att_mat[0]
    for i in np.arange(1,layers):
        joint_attentions[i] = aug_att_mat[i].dot(joint_attentions[i-1])

    return joint_attentions
```

```python
# ACKNOWLEDGEMENTS:
# Used Claude and ChatGPT to assist in writing the code
# Referenced tutorial notebooks from Lecture 6 ("CS263 Implementation Practice: Running and Training LLMs") as needed
# Referenced notebook provided by author of "Quantifying Attention Flow in Transformers" as needed for implementation and use of attention rollout: https://github.com/samiraabnar/attention_flow/

import torch
import matplotlib.pyplot as plt
from transformers import AutoTokenizer, AutoModelForCausalLM

# -------------------------------------------------
# Load model on GPU
# -------------------------------------------------

model_id = "meta-llama/Llama-3.2-1B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    attn_implementation="eager",
    device_map="auto",
)
model.eval()

for scenario in scenarios:
    for pronoun, sentence in scenario.items():

        # -------------------------------------------------
        # Build prompt
        # -------------------------------------------------

        messages = [
            {
                "role": "system",
                "content": (
                    "You are an emotion classifier. "
                    "Respond with only one word from this list ONLY: "
                    "anger, disgust, fear, joy, sadness, surprise. "
                    "DO NOT USE ANY OTHER WORD THAT IS NOT IN THIS LIST."
                ),
            },
            {"role": "user", "content": sentence},
        ]

        input_text = tokenizer.apply_chat_template(
            messages,
            add_generation_prompt=True,
            tokenize=False
        )

        # -------------------------------------------------
        # Find token span of scenario sentence
        # -------------------------------------------------

        # Tokenize just the scenario sentence
        sentence_tokens = tokenizer(sentence, add_special_tokens=False)["input_ids"]

        # Tokenize the entire input
        all_tokens = tokenizer(input_text)["input_ids"]

        # Find where the tokenized scenario sentence is in the input
        start, end = None, None
        for i in range(len(all_tokens)):
            if all_tokens[i : i + len(sentence_tokens)] == sentence_tokens:
                start = i
                end = i + len(sentence_tokens)
                break

        # -------------------------------------------------
        # Generate and get attentions
        # -------------------------------------------------

        inputs = tokenizer.apply_chat_template(
            messages,
            add_generation_prompt=True,
            tokenize=True,
            return_dict=True,
            return_tensors="pt",
        ).to(model.device)

        with torch.no_grad():
            hf_out = model.generate(
                **inputs,
                max_new_tokens=3,     # some words may get split into multiple tokens, so set to 3 instead of 1 to accommodate for this
                output_attentions=True,
                return_dict_in_generate=True,
            )

        # -------------------------------------------------
        # Decode generated word
        # -------------------------------------------------

        generated_ids = hf_out.sequences[0][inputs["input_ids"].shape[1]:]
        generated_word = tokenizer.decode(generated_ids, skip_special_tokens=True).strip()

        # -------------------------------------------------
        # Attention rollout (as defined in the paper "Quantifying Attention Flow in Transformers")
        # -------------------------------------------------

        step_attentions = hf_out.attentions[0]  # first generated token
        # Note: The word outputted by the model may be broken into multiple tokens.
        # For looking at and analyze attention, we look at the first token that is part of that word.

        # Stack layers, squeeze batch, mean over heads -> [layers, src_seq, src_seq]
        # (Paper explains one approach with multiple heads is to just average them)
        att_mat = torch.stack([a.squeeze(0).mean(dim=0) for a in step_attentions]).float().cpu().numpy()

        # Attention rollout (Abnar & Zuidema, 2020)
        joint_attentions = compute_joint_attention(att_mat)

        # Last layer, last row = generated token attending to all source tokens
        influence = torch.tensor(joint_attentions[-1, -1, :])  # [src_len]

        # -------------------------------------------------
        # Slice to scenario sentence tokens only
        # -------------------------------------------------

        sentence_influence = influence[start:end].unsqueeze(0).detach().cpu()
        tokens = [t.replace("Ġ", " ").strip() for t in tokenizer.convert_ids_to_tokens(all_tokens[start:end])]

        vmin, vmax = sentence_influence.min().item(), sentence_influence.max().item()
        if vmax - vmin < 1e-9:
            vmax = vmin + 1e-9

        # -------------------------------------------------
        # Plot
        # -------------------------------------------------

        plt.figure(figsize=(max(8, len(tokens) * 0.7), 2.5))

        plt.imshow(
            sentence_influence.numpy(),
            aspect="auto",
            cmap="Reds",
            vmin=vmin,
            vmax=vmax,
        )

        plt.xticks(
            range(len(tokens)),
            tokens,
            rotation=45,
            ha="right",
            fontsize=11,
        )

        plt.yticks([0], [generated_word], fontsize=11)
        plt.colorbar(label="Attention rollout")
        plt.title(f"Attention Rollout [{pronoun}]", fontsize=13)
        plt.tight_layout()
        plt.show()
```
